## Why This Matters
Every deep learning model — no matter how complex — is built on linear algebra.
- Model **weights** are matrices
- Input **data** (images, text, tabular) becomes tensors
- A neuron's output is literally a **dot product** + bias
- Backpropagation is chain-rule calculus applied to matrix operations

Understanding this layer means you'll actually understand *why* deep learning works, not just how to call `.fit()`.

## 1. Scalars, Vectors, Matrices, Tensors

| Object | Dimensions | Example | Used for in DL |
|---|---|---|---|
| Scalar | 0-D | `5` | A single weight, a loss value |
| Vector | 1-D | `[1, 2, 3]` | A feature row, a word embedding |
| Matrix | 2-D | `[[1,2],[3,4]]` | A layer's weights, a batch of tabular data |
| Tensor | N-D | shape `(batch, H, W, channels)` | A batch of images, a batch of sequences |

````python
# Scalar
scalar = 5

# Vector (1D)
vector = np.array([1, 2, 3])

# Matrix (2D)
matrix = np.array([[1, 2], [3, 4]])

# Tensor (3D) - e.g. a tiny batch of 2 "images", each 3x4
tensor = np.random.rand(2, 3, 4)

print("Scalar:", scalar, "| type:", type(scalar))
print("Vector:", vector, "| shape:", vector.shape, "| ndim:", vector.ndim)
print("Matrix:\n", matrix, "\n shape:", matrix.shape, "| ndim:", matrix.ndim)
print("Tensor shape:", tensor.shape, "| ndim:", tensor.ndim)
````

## 2. Vector Operations

### Addition
Element-wise. Vectors must be the same shape.

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b (element-wise) =", a * b)
````

### Dot Product
This is THE most important operation in deep learning.

$$a \cdot b = \sum_i a_i b_i$$

A single neuron computes: `output = dot(inputs, weights) + bias`

````python
dot_product = np.dot(a, b)
print("Dot product:", dot_product)

# Manual verification
manual = sum(a[i] * b[i] for i in range(len(a)))
print("Manual calc:", manual)

# Simulate a single neuron
inputs = np.array([0.5, 0.3, 0.9])
weights = np.array([0.2, 0.8, 0.1])
bias = 0.1
neuron_output = np.dot(inputs, weights) + bias
print("Neuron output:", neuron_output)
````

### Vector Norm
The "length" or magnitude of a vector. Used in regularization (L1/L2) and normalization.

````python
v = np.array([3, 4])
l2_norm = np.linalg.norm(v)          # Euclidean norm (sqrt of sum of squares)
l1_norm = np.linalg.norm(v, ord=1)   # Manhattan norm (sum of absolute values)

print("L2 norm:", l2_norm)  # sqrt(3^2 + 4^2) = 5
print("L1 norm:", l1_norm)  # |3| + |4| = 7
````

---
## 3. Matrix Operations

### Matrix Multiplication
This is what happens at every layer of a neural network: `output = matmul(inputs, weights) + bias`

````python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A @ B (matrix mult):\n", A @ B)          # preferred modern syntax
print("np.matmul(A, B):\n", np.matmul(A, B))    # same result
print("A * B (element-wise, NOT matrix mult):\n", A * B)  # common beginner mistake
````

### Transpose
Flips rows and columns. Used constantly in backpropagation.

````python
print("A:\n", A)
print("A transpose:\n", A.T)
````

### Inverse
Used in solving linear systems (e.g. closed-form linear regression: `w = (X^T X)^-1 X^T y`).

````python
A_inv = np.linalg.inv(A)
print("A inverse:\n", A_inv)

# Verify: A @ A_inv should be the identity matrix
print("A @ A_inv (should be identity):\n", np.round(A @ A_inv, 5))
````

---
## 4. Broadcasting
NumPy automatically expands smaller arrays to match shapes during operations — this is how bias gets added to every row in a batch without writing a loop.

````python
batch_inputs = np.array([[1, 2], [3, 4], [5, 6]])  # 3 samples, 2 features each
bias_vector = np.array([10, 20])                    # broadcast across every row

result = batch_inputs + bias_vector
print("Batch inputs:\n", batch_inputs)
print("Broadcasted result:\n", result)
````

## 5. The Real DS Bridge: Pandas → NumPy

In practice, your raw data lives in a Pandas DataFrame. Before it reaches a model, it gets converted into NumPy arrays (or PyTorch/TensorFlow tensors, which behave the same way).

````python
# Simulate a small tabular dataset
df = pd.DataFrame({
    'age': [25, 32, 47, 51],
    'income': [50000, 64000, 120000, 98000],
    'purchased': [0, 1, 1, 0]
})
print("DataFrame:\n", df)

# Convert features to NumPy array (this is what a model actually trains on)
X = df[['age', 'income']].to_numpy()
y = df['purchased'].to_numpy()

print("\nFeature matrix X:\n", X)
print("X shape:", X.shape)
print("Target vector y:", y)
````

**Key takeaway:** Pandas is for exploring/cleaning data. NumPy (or tensor libraries) is for the actual math a model runs on. Every DL pipeline follows this pattern:

```
Pandas (load/clean/explore) → NumPy/Tensor (convert) → Model (train)
```
